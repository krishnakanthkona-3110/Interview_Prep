# 1) what are P95 and P99 latency?

P95 and P99 latency are percentile-based performance metrics used to measure how fast a system responds.

# Why Not Just Use Average Latency?

Suppose an API response times are:

100ms, 120ms, 110ms, 130ms, 5000ms

Average:

1092ms

This doesn't represent most users' experience.

Percentiles help us understand the tail latency.

#P95 Latency

P95 means:

95% of requests completed within this latency.

Only:

5% of requests

were slower.

Example

Suppose 100 requests are sorted by latency.

The 95th request takes:

300 ms

Then:

P95 = 300 ms

Meaning:

95 requests <= 300 ms

5 requests > 300 ms

# P99 Latency

P99 means:

99% of requests completed within this latency.

Only:

1% of requests

were slower.

Example

If the 99th request takes:

800 ms

Then:

P99 = 800 ms

Meaning:

99 requests <= 800 ms
1 request > 800 ms

# 2) Guardrails:

guardrails are validation and governance layers that control the behavior of LLMs before, during, and after generation. Their purpose is to ensure that the model produces safe, reliable, and policy-compliant responses.

Guardrails can be implemented at multiple stages:

1) Input Guardrails – Validate user prompts, detect harmful content, PII, prompt injection, or jailbreak attempts before sending requests to the LLM.

2) Generation Guardrails – Apply constraints during generation, such as restricting topics, enforcing response formats, or limiting tool usage.

3) Output Guardrails – Validate the generated response for toxicity, hallucinations, sensitive information, compliance violations, or formatting requirements before returning it to the user.

# Matryoshka embeddings

 are embeddings trained so that meaningful semantic information is preserved even when the vector is truncated to lower dimensions. This allows the same embedding to be used at different dimensionalities, such as 1024, 512, or 256 dimensions, enabling a trade-off between retrieval accuracy, storage requirements, and search latency. In RAG systems, this helps optimize vector database performance and cost without needing separate embedding models for different use cases.


 A Matryoshka embedding is trained so that the first part of the embedding vector contains the most important information, and you can truncate the vector to smaller dimensions while still retaining useful semantic meaning.

We perform cross validation (to make sure that model has good accuracy rate and it can be used for prediction using unseen/new or test data). To do so, we use train and test data by properly splitting our dataset for example 80% for training, 20% for testing the model. This can be performed using train_test, train_test_split or K-fold (K-fold is mostly used to avoid under and overfiting problems).

A model is considered as a good model when it gives high accuracy using training as well as testing data. Good accuracy on test data means, model will have good accuracy when it is trying to make predictions on new or unseen data for example, using the data which is not included in the training set.

Good accuracy also means that the value predicted by the model will be very much close to the actual value.

Bias will be low and variance will be high when model performs well on the training data but performs bad or poorly on the test data. High variance means the model cannot generalize to new or unseen data. (This is the case of overfiting)

If the model performs poorly (means less accurate and cannot generalize) on both training data and test data, it means it has high bias and high variance (Low Variance). (This is the case of underfiting)

If model performs well on both test and training data. Performs  well meaning, predictions are close to actual values for unseens data so accuracy will be high. In this case, bias will be low and variance will also be low.

The best model must have low bias (low error rate on training data) and low variance (can generalize and has low error rate on new or test data).

(This is the case for best fit model) so always have low bias and low variance for your models.

# Performance Metrics:

# Confusion Matrix

A confusion matrix is a table used to understand the performance of a classification model. It compares the actual values with the values predicted by the model . For a binary classification problem (with classes 0 and 1), it is a 2x2 matrix .

- True Positive (TP): The model correctly predicts the positive class (Actual: 1, Predicted: 1) .

- True Negative (TN): The model correctly predicts the negative class (Actual: 0, Predicted: 0) .

- False Positive (FP): The model incorrectly predicts the positive class when it is actually negative (Actual: 0, Predicted: 1). This is also known as a Type I error .

- False Negative (FN): The model incorrectly predicts the negative class when it is actually positive (Actual: 1, Predicted: 0). This is also known as a Type II error .

The main goal is to maximize True Positives and True Negatives while minimizing False Positives and False Negatives .

# Accuracy

Accuracy is the most intuitive performance measure. It is the ratio of correctly predicted observations to the total observations .

Formula:

Accuracy=

TP+TN / TP+TN+FP+FN


However, accuracy is not a good metric for imbalanced datasets. An imbalanced dataset has a significant disparity between the number of samples in different classes. For example, if a dataset has 900 samples of "Class 0" and 100 of "Class 1," a model that always predicts "Class 0" would achieve 90% accuracy but would be useless for identifying "Class 1" .

#Precision and Recall

For imbalanced datasets, precision and recall are more insightful metrics .

# Precision

Precision answers the question: "Out of all the positive predictions made by the model, how many were actually correct?" It focuses on minimizing False Positives .

Formula:

Precision=
TP / TP+FP


When to use: Precision is important when the cost of a False Positive is high.

Example (Spam Detection): If a non-spam email (actual negative) is classified as spam (predicted positive), it's a False Positive. This is a critical error as an important email might be missed. Therefore, high precision is required .

# Recall

Recall (also known as Sensitivity or True Positive Rate) answers the question: "Out of all the actual positive cases, how many did the model correctly identify?" It focuses on minimizing False Negatives .

Formula:

Recall=
TP /TP+FN


When to use: Recall is important when the cost of a False Negative is high.

Example (Cancer Detection): If a person who has cancer (actual positive) is diagnosed as not having cancer (predicted negative), it's a False Negative. This is a life-threatening error. Therefore, high recall is crucial .

# Ensemble Learning:
Ensemble Learning is a machine learning technique where multiple models (called base learners or weak learners) are combined to produce a better prediction than any single model.

Why use Ensemble Learning?

- Improves accuracy
- Reduces overfitting
- Increases robustness
- Produces more stable predictions

# Bagging (Bootstrap Aggregating):

Bagging is an ensemble learning technique where multiple models are trained independently and in parallel on bootstrap samples of the training data.

The predictions from all models are then combined using majority voting for classification or averaging for regression to produce the final prediction. Bagging primarily helps reduce variance and improve model stability.

The main goal of Bagging is to reduce variance and prevent overfitting and improve generalization.

- Bootstrap samples: Creating multiple subsets of the training dataset by randomly sampling data points with replacement. Because sampling is done with replacement, the same data point can appear multiple times in a subset, while some data points may not appear at all.

-- Here, we sample both rows as well as columns

- Aggregation: Combining the predictions of all models, typically using majority voting for classification problems or averaging for regression problems.

Hence Bagging is also called as Bootstrap Aggregation.

# Boosting:

Booasting is an ensemble learning method, that improves accuracy by training models one after another.

Boosting is an ensemble learning technique where multiple weak learners are trained sequentially. Each new model focuses on correcting the errors made by the previous models. The predictions of all models are then combined, typically using weighted voting for classification or weighted averaging for regression, to create a stronger and more accurate model.

#"Why do we use Boosting?"

Boosting primarily helps reduce bias and improve predictive accuracy by making each new model learn from the mistakes of the previous ones.

#"How is it different from Bagging?"

In Bagging, models are trained independently and in parallel on bootstrap samples to reduce variance.
In Boosting, models are trained sequentially, where each model learns from the errors of previous models to reduce bias.

Popular boosting algorithms include AdaBoost, XGBoost, and LightGBM.

# Steps to perform Boosting:

- Start with Equal Weights: The algorithm gives equal importance to all data points and trains the first base model (usually a decision tree)

- Identify Errors: It checks which predictions were wrong and increase the weight of those samples, making them more important for the next model.

- Train the Next Model: A new model is trained with the updated weights, focusing more on the difficult cases.

- Repeat Until Optimized: This process continues until the error rate is low or the set number of iterations is reached.

# Advantages of Boosting in ML.

- Higher Accuracy:

Boosting combines multiple weak learners to create a strong learner, which improves the overall prediction accuracy.

- Less Overfitting:

It reduces the risk of overfitting by focusing on difficult data points and balancing errors.

Boosting focuses on reducing errors from previous models and can improve generalization, especially when properly tuned. (However, too many boosting iterations can still cause overfitting.)

- Handles Imbalanced Data:

Boosting can give more importance to misclassified samples, so it can focus on difficult or minority class examples and perform better on imbalanced datasets.

- Better Interpretability:

Since Boosting builds models step by step, it helps in understanding how decisions are made.


Many boosting algorithms provide feature importance scores, which help understand which features contribute most to the prediction.

# 1. AdaBoost (Adaptive Boosting)
- Trains models sequentially and increases the weight of misclassified samples after each iteration.
- The next model focuses more on the mistakes made by previous models

Key Idea: Focus on misclassified data points.

# 2. Gradient Boosting
- Instead of adjusting sample weights, each new model learns the residual errors (mistakes) of the previous model.
- Uses gradient descent principles to minimize the loss function.

Key Idea: Fit the errors of previous models.

# 3. XGBoost
- An optimized version of Gradient Boosting.
- Adds regularization, parallel processing, and efficient handling of missing values, making it faster and less prone to overfitting.

Key Idea: Faster, more accurate, and less prone to overfitting.

# 4. LightGBM
- A faster and more memory-efficient implementation of Gradient Boosting.
- Uses leaf-wise tree growth, making it suitable for large datasets.

Key Idea: Faster training and lower memory usage.

# 5. CatBoost
- Designed to handle categorical features automatically.
- Requires less preprocessing and helps reduce overfitting.
- Reduces overfitting and prediction shift.

Key Idea: Best when data contains many categorical variables.
