# RAG vs Fine-tuning vs Prompt Engineering

I use RAG when the model needs access to external, private, large, or frequently changing knowledge. I use fine-tuning when I need to change the model's behavior, style, or performance on a specific task. RAG updates knowledge without retraining, while fine-tuning changes the model itself. In some systems, I use both together."
