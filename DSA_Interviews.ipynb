# Binary Search

```python
def binary_search(arr,k):
  n = len(arr)
  low = 0
  high = len(arr)-1

  while low<=high:
    mid = (low+high)//2

    if arr[mid]==k:
      return mid

    elif arr[mid] >k:
      high = mid-1
    else:
      low = mid +1
  return -1

a = [2,4,5,6,8,9,14]
```

# Print in the decreasing order of frequency

```python
a = [2,6,3,8,2,8,2,3,8,10,6]
dic = {}
for i in a:
  if i in dic:
    dic[i] +=1
  else:
    dic[i] = 1
dic
k = sorted(dic.items(),key = lambda x:x[1], reverse=True)
d = []
for i,j in k:
  d.extend([i]*j)
d
```

# sort the list without using sort()

```python
arr = [2,6,3,8,2,8,2,3,8,10,6,9]

n = len(arr)
for i in range(n):
  for j in range(1,n-i):
    if arr[j]<arr[j-1]:
      arr[j],arr[j-1] = arr[j-1],arr[j]
arr

```

# Palindrome

```python
num = "121"

left = 0
right = len(num) - 1

while left < right:
    if num[left] != num[right]:
        print("Not Palindrome")
        break

    left += 1
    right -= 1
else:
    print("Palindrome")
```
