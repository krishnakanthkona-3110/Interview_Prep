#1) Binary Search

```python
def binary_search(arr,k):
  n = len(arr)
  low = 0
  high = len(arr)-1

  while low<=high:
    mid = (low+high)//2

    if arr[mid]==k:
      return mid

    elif arr[mid] >k:
      high = mid-1
    else:
      low = mid +1
  return -1

a = [2,4,5,6,8,9,14]
```

```python
# Optimized solution
def quick_sort(arr):
  if len(arr) <= 1:
    return arr

  p = arr[-1]

  left = [x for x in arr[:-1] if x <= p]
  right = [x for x in arr[:-1] if x > p]

  left = quick_sort(left)
  right = quick_sort(right)

  return left + [p] + right
quick_sort([2,6,3,8,2,8,2,3,8,10,6,9])
```

#12) Two sum

```python
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]

nums = [2, 7, 11, 15]
target = 9

print(two_sum(nums, target))
```

```python
# Optimized approach
def two_sum(nums, target):
    seen = {}

    for i in range(len(nums)):
        complement = target - nums[i]

        if complement in seen:
            return [seen[complement], i]

        seen[nums[i]] = i

    return []

nums = [2, 7, 11, 15]
target = 9

print(two_sum(nums, target))
```

# 15) Max sub array

```python
def max_subarray(nums):
    n = len(nums)
    max_sum = float('-inf')

    for i in range(n):
        current_sum = 0

        for j in range(i, n):
            current_sum += nums[j]
            max_sum = max(max_sum, current_sum)

    return max_sum


nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

print(max_subarray(nums))
```

# 16) Valid Parentheses

```python
# greg video (in case of doubt)
def valid(str):
  dic = {')':'(','}':'{',']':'['}
  stack = []

  for i in str:
    # if it is a open bracket
    if i not in dic:
      stack.append(i)
    else:# closed bracket
      if not stack: # if stack empty
        return False
      else:
        popped = stack.pop() # most recent bracket in stack
        if popped != dic[i]: # if last value in stack not equal with current value in str
          return False
  return len(stack)==0
```

# 17) Longest Palindrome Substring

```python
def longest_palindromic_substring_bruteforce(s):
    n = len(s)
    max_len = 0

    # Step 1: Generate all substrings
    for i in range(n):
        for j in range(i, n):
            substring = s[i:j+1]

            # Step 2: Check if substring is palindrome
            if substring == substring[::-1]:
                # Step 3: Update max length
                max_len = max(max_len, len(substring))

    return max_len

print(longest_palindromic_substring_bruteforce("babad"))  # Output: 3
```

```python
# Optimized solution
def longestPalindrome(s):
    if not s:
        return ""

    start = 0
    end = 0

    for i in range(len(s)):

        # Odd-length palindrome
        left = i
        right = i

        while left >= 0 and right < len(s) and s[left] == s[right]:
            if right - left > end - start:
                start = left
                end = right

            left -= 1
            right += 1

        # Even-length palindrome
        left = i
        right = i + 1

        while left >= 0 and right < len(s) and s[left] == s[right]:
            if right - left > end - start:
                start = left
                end = right

            left -= 1
            right += 1

    return s[start:end + 1]
```

# 18) Reverse a linked list

```python
def printall(head):
  curr = head
  prev = None

  while curr:
    next_node = curr.next
    curr.next = prev
    prev = curr
    curr = next_node
  return prev
```

I'll use three pointers: prev, curr, and next_node. prev keeps the reversed portion, curr is the current node, and next_node saves the remaining list before I change the pointer.

# 19) Merge Two Sorted Arrays

```python
def mergesorted(arr1,arr2):
  n1 = len(arr1)
  n2 = len(arr2)
  i = 0
  j = 0

  res = []
  while i < n1 and j < n2:
    if arr1[i]<arr2[j]:
      res.append(arr1[i])
      i += 1
    else:
      res.append(arr2[j])
      j += 1

  while i<n1:
    res.append(arr1[i])
    i += 1

  while j < n2:
    res.append(arr2[j])
    j += 1

  return res

arr1 = [1, 3, 5, 7]
arr2 = [2, 4, 6, 8]
print(mergesorted(arr1,arr2))
```

# 21) Stairs
You are climbing a staircase and it takes A steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

Return the number of distinct ways modulo

```python
def stair(n):
  if n==1:
    return 1
  if n==2:
    return 2
  return stair(n-1)+stair(n-2)
stair(5)
```

# 22) Coins
You are given:
coins = [c1, c2, c3...]

amount

Find the minimum number of coins required to make the given amount. If not possible → return -1.

```python
def coin_change(coins, amount):

    dp = [float('inf')] * (amount + 1)

    dp[0] = 0

    for coin in coins:

        for current_amount in range(coin, amount + 1):

            dp[current_amount] = min(
                dp[current_amount],
                dp[current_amount - coin] + 1
            )

    if dp[amount] == float('inf'):
        return -1

    return dp[amount]


coins = [1, 3, 4]
amount = 6

print(coin_change(coins, amount))
```
