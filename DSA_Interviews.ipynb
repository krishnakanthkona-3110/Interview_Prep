#1) Binary Search

```python
def binary_search(arr,k):
  n = len(arr)
  low = 0
  high = len(arr)-1

  while low<=high:
    mid = (low+high)//2

    if arr[mid]==k:
      return mid

    elif arr[mid] >k:
      high = mid-1
    else:
      low = mid +1
  return -1

a = [2,4,5,6,8,9,14]
```

#5) sort the list without using sort()

```python
arr = [2,6,3,8,2,8,2,3,8,10,6,9]

n = len(arr)
for i in range(n):
  for j in range(1,n-i):
    if arr[j]<arr[j-1]:
      arr[j],arr[j-1] = arr[j-1],arr[j]
arr
```

```python
# Optimized solution
def quick_sort(arr):
  if len(arr) <= 1:
    return arr

  p = arr[-1]

  left = [x for x in arr[:-1] if x <= p]
  right = [x for x in arr[:-1] if x > p]

  left = quick_sort(left)
  right = quick_sort(right)

  return left + [p] + right
quick_sort([2,6,3,8,2,8,2,3,8,10,6,9])
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


print(factorial(5))
```

#8) Palindrome

```python
num = "12521"

left = 0
right = len(num) - 1

while left < right:
    if num[left] != num[right]:
        print("Not Palindrome")
        break

    left += 1
    right -= 1
else:
    print("Palindrome")
```

#10) Longest substring without repeating characters

```python
def longest_substring(s):
    char_index = {}
    left = 0
    max_length = 0

    for right in range(len(s)):

        if s[right] in char_index and char_index[s[right]] >= left:
            left = char_index[s[right]] + 1

        char_index[s[right]] = right
        #print(char_index)
        max_length = max(max_length, right - left + 1)

    return max_length


s = "abcabcbb"

print(longest_substring(s))
```

#11) Two sum

```python
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]

nums = [2, 7, 11, 15]
target = 9

print(two_sum(nums, target))
```

```python
# Optimized approach
def two_sum(nums, target):
    seen = {}

    for i in range(len(nums)):
        complement = target - nums[i]

        if complement in seen:
            return [seen[complement], i]

        seen[nums[i]] = i

    return []

nums = [2, 7, 11, 15]
target = 9

print(two_sum(nums, target))
```

# 12) Move Zeroes - Move all 0s to the end of the array while maintaining the order of the non-zero elements.

```python
nums = [0, 1, 0, 3, 12]

non_zero = []

for x in nums:
    if x != 0:
        non_zero.append(x)

zeroes = len(nums) - len(non_zero)

nums = non_zero + [0] * zeroes

print(nums)
```

```python
nums = [0, 1, 0, 3, 12]

j = 0

for i in range(len(nums)):
    if nums[i] != 0:
        nums[i], nums[j] = nums[j], nums[i]
        j += 1

print(nums)
```

# 13) Find the k most frequent elements in a list.

```python
nums = [1, 1, 1, 2, 2, 3]
k = 2

freq = {}

for num in nums:
  if num in freq:
    freq[num] += 1
  else:
    freq[num] = 1

sorted_freq = sorted(freq, key=freq.get, reverse=True)

print(freq)
print(sorted_freq[:k])
```
