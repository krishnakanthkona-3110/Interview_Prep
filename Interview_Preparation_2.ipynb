Neural Network → RNN → LSTM/GRU → Transformer → Large Language Models (GPT, Claude, Gemini) → Agentic AI Systems.

Traditional Neural Networks process inputs independently and have no memory of previous inputs. This works well for structured data but not for sequential data like text or speech. To handle sequences, RNNs were introduced. RNNs maintain a hidden state that carries information from previous time steps, enabling them to understand context. However, RNNs struggle with long-term dependencies due to vanishing gradients, which led to the development of LSTM and GRU architectures. Later, Transformers replaced RNNs for many NLP tasks because they can capture long-range relationships more efficiently using self-attention.

# RNN :-

RNN is a type of neural network designed for sequential data where the output depends not only on the current input but also on previous inputs.

# Self Attention:

Self-Attention is a mechanism in Transformers where every word in a sequence attends to every other word in the same sequence. Queries, Keys, and Values are all derived from the same input. This helps the model capture contextual relationships and dependencies, regardless of the distance between words.

# Cross Attention -

It is a mechanism that lets one sequence interrogate another. It is a secret ingredient behind translation, image generation and much more.

Cross-Attention is an attention mechanism where the Query (Q) comes from one sequence, while the Key (K) and Value (V) come from a different sequence. It allows one sequence to attend to and gather relevant information from another sequence.

Cross-Attention is a mechanism used mainly in Encoder-Decoder Transformers where the decoder attends to the encoder's output. The Query comes from the decoder, while the Keys and Values come from the encoder. This helps the decoder focus on the most relevant parts of the input sequence while generating each output token.

Self-Attention learns relationships within a sequence, whereas Cross-Attention learns relationships between two different sequences.
