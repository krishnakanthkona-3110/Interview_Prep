Quantization is the process of reducing the numerical precision of model parameters, activations, or both, to make the model smaller, faster, and more memory-efficient, while trying to minimize the loss in accuracy.


There are two major approaches: Post-Training Quantization, where an already-trained model is quantized, and Quantization-Aware Training, where the model learns to handle quantization during training.

For LLMs, popular PTQ methods include GPTQ and AWQ. We can also classify quantization as weight-only or weight-plus-activation, and as static or dynamic quantization. GPTQ is primarily a weight-only PTQ method, while QLoRA combines 4-bit quantization with LoRA for parameter-efficient fine-tuning.”

Quantization is the process of reducing the numerical precision of a model's weights (and sometimes activations) — e.g., from 32-bit floating point (FP32) down to 16-bit, 8-bit, or even 4-bit representations — to shrink model size, cut memory bandwidth, and speed up inference, with a controlled trade-off in accuracy.


to reduce the computational complexity of the model and improve its efficiency during training and inference.

Quantization aims to reduce the precision of a models parameters from higher bit-widths (like 32-bit floating
point) to lower bit-widths (like 8-bit integers).

There is often some loss of precision when reducing the number of bits.

The main goal of quantization is to reduce the number of bits needed to represent the original parameters while preserving their precision as best as possible.


Why do we need quantization?

For LLMs, the main benefits are:

- Reduce model size
- Reduce GPU/CPU memory consumption
- Improve inference speed
- Reduce bandwidth requirements
- Make large models possible to run on consumer GPUs / edge devices
