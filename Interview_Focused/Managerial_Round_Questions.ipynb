# How Do You Handle Pressure?

Don't say:

"I work very well under pressure."

Instead give a process.

"When there is pressure, I first prioritize tasks based on business impact and urgency. For a production issue, for example, my first priority is restoring the service or minimizing the impact. Then I investigate the root cause. I also communicate the status clearly to the stakeholders rather than waiting until the issue is completely resolved.

Once the immediate issue is handled, I focus on the root-cause fix and preventive measures."

That's a manager-level answer.

# What If You Disagree With Your Manager?

I try to treat disagreement as a signal that we have different information, not that someone's wrong. My approach is: understand their reasoning first, bring data to back my view, and once a decision is made — commit to it fully.

For example, while building a multi-agent RAG platform for querying financial and operational reports, there was a design discussion around how to handle document retrieval for image-heavy financial statements. One approach favored a simpler text-only embedding pipeline for speed and lower infra cost. I felt that would lose accuracy on documents where the layout and visuals carried the information — tables, charts — so I proposed a vision-based retrieval approach that could embed page images directly instead of relying purely on extracted text.

Rather than just push back, I laid out the tradeoff clearly — the retrieval accuracy metrics we'd expect to gain (recall@k, precision) versus the added latency and infra cost — and suggested we validate it on a small sample of documents before committing either way. That data-first framing made it a shared decision instead of a standoff, and we moved forward with the vision-based approach after the sample confirmed the accuracy gap was real.

The bigger principle for me: I raise disagreement early and directly with data, but I don't let it become personal or repeated litigation. Once the team or manager decides, I execute on it fully — even if it's not exactly what I proposed — because alignment matters more than being right."

Very common.

If the final decision is still different from my recommendation, I would align with the decision and focus on executing it effectively, unless there is a significant technical or security risk that needs to be escalated."

Excellent managerial-round answer.

# "How Do You Handle a Mistake?"

Never say:

"I don't make mistakes."

Say:

"If I make a mistake, my first priority is to understand the impact and communicate it appropriately. I would take ownership, fix the immediate problem, identify the root cause and then put a preventive measure in place, such as validation, monitoring or an additional test case.

I believe the important thing is not hiding a mistake but making sure the same issue doesn't happen again."

# How Do You Prioritize Multiple Tasks?

Use:

Impact + Urgency + Dependency + Effort

"I prioritize based on business impact and urgency first. Production issues and customer-impacting problems take priority. Then I consider dependencies—if one task is blocking multiple team members, I prioritize that accordingly. For normal development work, I consider deadlines and effort and communicate if priorities conflict."
