# Prompt Engineering

Prompt engineering is the process of designing and optimizing prompts to guide an LLM to produce accurate, relevant, consistent, and desired outputs for a specific task.

It involves clearly defining the instruction, providing necessary context, setting constraints, giving examples when required, and specifying the expected output format.

Common techniques include zero-shot, few-shot, role prompting, structured output, and context-based prompting. In my RAG projects, I mainly use instruction prompting, context injection, and constraints to ensure the LLM answers based on the retrieved context.


Prompt Engineering = Giving the LLM the right instructions, context, examples, and constraints to get the desired output.

For example, instead of:

- Explain this document.

we can write:

- You are a financial analyst.
Analyze the following document and summarize:
1. Revenue
2. Expenses
3. Profit
4. Key risks

Answer in a table and use only information present in the document.

The second prompt provides role + task + context + constraints + output format, making the response more controlled.

# Important Prompt Engineering Techniques:-

# 1. Zero-Shot Prompting

- Ask the model to perform a task without providing examples.

Classify the sentiment of this review as Positive, Negative, or Neutral.

Review:
"The product arrived late."

There are no examples.

When to use: Simple and well-defined tasks.

# 2. One-Shot Prompting

- Provide one example before asking the model to perform the task.

Example:
Review: "The product is excellent."
Sentiment: Positive

Now classify:
Review: "The product arrived late."
Sentiment:

The example demonstrates the expected behavior.

# 3. Few-Shot Prompting

- Provide multiple examples.

Review: "Excellent product."
Sentiment: Positive

Review: "Terrible quality."
Sentiment: Negative

Review: "It is okay."
Sentiment: Neutral

Now classify:
Review: "The product is not bad."
Sentiment:

Key interview point:

Few-shot prompting helps the model understand the expected input-output pattern through examples.

# 4. Role Prompting / Persona Prompting

- Assign a role to the LLM.

You are an experienced Python developer.
Review the following code and identify performance issues.

This provides additional behavioral/contextual guidance.

5. Instruction Prompting

- Clearly specify what the model should do.

Extract the customer's name, email, and order ID from the text.
Return the result as JSON.

Good instructions should be:

Clear,
Specific,
Unambiguous,
Action-oriented

# 6. Chain-of-Thought Prompting

- This encourages the model to reason through a problem step by step.

Conceptually:

Solve the problem step by step before providing the final answer.

For interviews, be careful with how you describe this:

Chain-of-thought prompting is a technique that encourages step-by-step reasoning for complex tasks.

# 7. Structured Output Prompting

- Tell the model exactly how the response should be formatted.

For example:

Return the result in this format:

{
    "name": "",
    "email": "",
    "sentiment": ""
}

This is especially useful in enterprise applications, where the LLM output is consumed by another application.

Modern systems can also use structured-output / JSON-schema mechanisms rather than relying only on prompt instructions.

# 8. Constraint-Based Prompting

- Specify what the model must or must not do.

Example:

Answer using only the provided context.
If the answer is not present, say:
"I don't have enough information."
Do not use outside knowledge.

This is particularly important in RAG.

# 9. Context Injection

- Provide relevant information to the model along with the instruction.

In RAG:

User Query
     ↓
Retriever
     ↓
Relevant Documents
     ↓
Prompt + Retrieved Context
     ↓
LLM

Example:

Answer the question using ONLY the following context.

Context:
{retrieved_chunks}

Question:
{user_question}

This is one of the most important techniques for RAG systems.

# 10. Prompt Chaining

- Instead of using one prompt, use the output of one prompt as the input to another.

This is useful when a task is too complex for a single prompt.

# 11. ReAct — Reason + Act

Very important for AI agents.

- ReAct combines reasoning with actions/tool usage:

Think → Act → Observe → Think → Act → ...

For example, an agent might:

User: What is the weather in Hyderabad?

Agent

 ↓

Decides it needs weather information

 ↓

Calls weather tool

 ↓

Observes result

 ↓

Generates final answer

#Interview definition:
ReAct is an agentic prompting approach where the model combines reasoning with actions, such as calling tools, and uses the observations from those actions to continue solving the task.

# Prompt Engineering vs Prompt Optimization

This is a useful interview distinction.

- Prompt Engineering:

Designing prompts to get desired model behavior.

- Prompt Optimization:

Systematically improving prompts based on evaluation results.
